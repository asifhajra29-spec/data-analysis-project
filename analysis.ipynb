# Retail Sales Analysis 📊

A beginner-friendly Business Analytics project exploring sales, profit, regions, categories, and discounts using Python.

## Business Questions
- Which product category generates the most sales and profit?
- Which region contributes the most sales?
- How do discounts relate to profit?
- How does sales performance change over time?

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data/sales_data.csv')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.head()
```

## Data Quality Check

```python
print('Shape:', df.shape)
print('\nMissing values:')
print(df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
print('\nData types:')
print(df.dtypes)
```

## Overall Performance

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
avg_order_value = df['Sales'].mean()

print(f'Total sales: ${total_sales:,.2f}')
print(f'Total profit: ${total_profit:,.2f}')
print(f'Average order value: ${avg_order_value:,.2f}')
```

## Sales & Profit by Category

```python
category_summary = df.groupby('Category')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
category_summary
```

```python
category_summary['Sales'].plot(kind='bar', title='Sales by Category')
plt.ylabel('Sales ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Regional Performance

```python
region_summary = df.groupby('Region')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
region_summary
```

```python
region_summary['Sales'].plot(kind='bar', title='Sales by Region')
plt.ylabel('Sales ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Monthly Sales Trend

```python
monthly_sales = df.set_index('Order_Date').resample('ME')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o', title='Monthly Sales Trend')
plt.ylabel('Sales ($)')
plt.xlabel('Month')
plt.tight_layout()
plt.show()
```

## Discount vs Profit

```python
discount_profit = df.groupby('Discount')['Profit'].mean().reset_index()
discount_profit
```

```python
print('Correlation between discount and profit:', round(df['Discount'].corr(df['Profit']), 3))

plt.scatter(df['Discount'], df['Profit'])
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit ($)')
plt.tight_layout()
plt.show()
```
